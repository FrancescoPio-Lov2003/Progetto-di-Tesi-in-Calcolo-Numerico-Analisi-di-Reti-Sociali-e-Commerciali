```python
import os
import time
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

def genera_grafo_scenario1_cs2(percorso_input, percorso_output):
    print(f"⚙️ Inizio generazione grafo: Scenario 1 (E-commerce - Acquisti Diretti)")
    inizio = time.time()
    
    # ===============================================
    # PASSO 1: CARICAMENTO DATASET E CREAZIONE GRAFO
    # ===============================================
    print(f"📥 Caricamento del dataset da: {percorso_input}...")
    df = pd.read_csv(percorso_input)
    
    # Crea il grafo orientato e rimuove eventuali auto-loop
    G = nx.from_pandas_edgelist(df, 'Source', 'Target', create_using=nx.DiGraph())
    G.remove_edges_from(list(nx.selfloop_edges(G)))

    # ===============================================
    # PASSO 2: DEFINIZIONE DEI GRUPPI STRATEGICI
    # ===============================================
    print("🧠 Classificazione dei nodi...")
    best_sellers = [0, 1, 2]
    
    # Nodi speciali (quelli che avranno i numeri stampati)
    nodi_speciali = best_sellers
    
    # Tutti gli altri nodi (Prodotti Isolati)
    prodotti_isolati = [n for n in G.nodes() if n not in nodi_speciali]

    # ===============================================
    # PASSO 3: CALCOLO DEL LAYOUT SPAZIALE
    # ===============================================
    print("🌐 Calcolo della disposizione spaziale (Layout espanso)...")
    plt.figure(figsize=(16, 12)) 
    
    # Stesso K, iterazioni e seed per massima coerenza visiva
    pos = nx.spring_layout(G, k=0.9, iterations=100, seed=42) 

    # ===============================================
    # PASSO 4: DISEGNO DEI NODI E DEGLI ARCHI
    # ===============================================
    print("🎨 Disegno della rete (Colori e Dimensioni)...")
    
    # Disegno le categorie con colori e dimensioni allineati al resto della tesi
    nx.draw_networkx_nodes(G, pos, nodelist=best_sellers, node_color='red', node_size=900, label='Best-Seller (0, 1, 2)')
    nx.draw_networkx_nodes(G, pos, nodelist=prodotti_isolati, node_color='skyblue', node_size=60, alpha=0.5, label='Prodotti Isolati (Acquisti Diretti)')

    # Archi leggeri
    nx.draw_networkx_edges(G, pos, alpha=0.15, edge_color='gray', arrows=True, arrowsize=12)

    # ===============================================
    # PASSO 5: ETICHETTE E DETTAGLI ESTETICI
    # ===============================================
    # Inserimento dei numeri SOLO per i nodi speciali (Best-Seller)
    labels = {n: str(n) for n in nodi_speciali}
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=12, font_weight='bold', font_color='black')

    plt.title("Caso di Studio 2: Acquisti Diretti e Prodotti Isolati (Scenario 1)", fontsize=18, fontweight='bold', pad=20)
    
    # Legenda esterna rimpicciolita
    plt.legend(scatterpoints=1, bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=12, markerscale=0.3)
    
    plt.margins(0)
    plt.axis('off')

    # ===============================================
    # PASSO 6: SALVATAGGIO AUTOMATICO
    # ===============================================
    cartella_destinazione = os.path.dirname(percorso_output)
    if not os.path.exists(cartella_destinazione):
        os.makedirs(cartella_destinazione)
        print(f"📁 Creata cartella: {cartella_destinazione}")

    print(f"💾 Salvataggio in corso...")
    plt.savefig(percorso_output, dpi=300, bbox_inches='tight')
    
    plt.close() 

    fine = time.time()
    print(f"✅ Grafo salvato con successo in: {percorso_output}")
    print(f"⏱️ Tempo impiegato: {fine - inizio:.2f} secondi!\n")


FILE_INPUT_S1 = '../DataSet_CasoStudio2/Rete_100/dataset_scenario1_cs2.csv'
FILE_OUTPUT_S1 = '../Grafi_CasoStudio2/Grafo_Scenario1_cs2.png'

genera_grafo_scenario1_cs2(FILE_INPUT_S1, FILE_OUTPUT_S1)
```

```python
import os
import time
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

def genera_grafo_scenario2_cs2(percorso_input, percorso_output):
    print(f"⚙️ Inizio generazione grafo: Scenario 2 (Rete Mista e Prodotti Hub)")
    inizio = time.time()
    
    # ===============================================
    # PASSO 1: CARICAMENTO DATASET E CREAZIONE GRAFO
    # ===============================================
    print(f"📥 Caricamento del dataset da: {percorso_input}...")
    df = pd.read_csv(percorso_input)
    
    # Crea il grafo orientato
    G = nx.from_pandas_edgelist(df, 'Source', 'Target', create_using=nx.DiGraph())
    G.remove_edges_from(list(nx.selfloop_edges(G)))

    # ===============================================
    # PASSO 2: DEFINIZIONE DEI GRUPPI STRATEGICI
    # ===============================================
    print("🧠 Classificazione dei nodi...")
    best_sellers = [0, 1, 2]
    hubs = [60, 70, 80, 90]
    
    # I prodotti isolati vanno dal nodo 3 al nodo 50
    prodotti_isolati = [n for n in G.nodes() if 3 <= n <= 50]
    
    # I prodotti correlati vanno dal 51 in poi, escludendo gli Hub
    prodotti_correlati = [n for n in G.nodes() if n >= 51 and n not in hubs]
    
    # Nodi speciali che avranno i numeri stampati sopra
    nodi_speciali = best_sellers + hubs

    # ===============================================
    # PASSO 3: CALCOLO DEL LAYOUT SPAZIALE (OTTIMIZZATO)
    # ===============================================
    print("🌐 Calcolo della disposizione spaziale (Layout espanso)...")
    plt.figure(figsize=(16, 12)) 
    
    # MODIFICA: k aumentato a 0.85 per amplificare la forza di repulsione e distanziare i nodi Hub (60, 70, 80, 90)
    pos = nx.spring_layout(G, k=0.85, iterations=120, seed=42) 

    # ===============================================
    # PASSO 4: DISEGNO DEI NODI E DEGLI ARCHI
    # ===============================================
    print("🎨 Disegno della rete (Allineamento colori e legenda aggiornata)...")
    
    # Disegno le categorie assegnando i colori definiti con l'aggiunta degli intervalli nella legenda
    nx.draw_networkx_nodes(G, pos, nodelist=best_sellers, node_color='red', node_size=900, label='Best-Seller (0, 1, 2)')
    nx.draw_networkx_nodes(G, pos, nodelist=hubs, node_color='orange', node_size=700, label='Prodotti Hub / Civetta (60, 70, 80, 90)')
    nx.draw_networkx_nodes(G, pos, nodelist=prodotti_correlati, node_color='lightgreen', node_size=60, alpha=0.6, label='Prodotti Correlati (51-99)')
    nx.draw_networkx_nodes(G, pos, nodelist=prodotti_isolati, node_color='skyblue', node_size=60, alpha=0.6, label='Prodotti Isolati (3-50)')

    # Archi leggeri
    nx.draw_networkx_edges(G, pos, alpha=0.15, edge_color='gray', arrows=True, arrowsize=12)

    # ===============================================
    # PASSO 5: ETICHETTE E DETTAGLI ESTETICI
    # ===============================================
    labels = {n: str(n) for n in nodi_speciali}
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=12, font_weight='bold', font_color='black')

    plt.title("Caso di Studio 2: Rete Mista e Prodotti Hub (Scenario 2)", fontsize=18, fontweight='bold', pad=20)
    
    # Legenda esterna rimpicciolita
    plt.legend(scatterpoints=1, bbox_to_anchor=(1.02, 1), loc='upper left', fontsize=12, markerscale=0.3)
    
    plt.margins(0)
    plt.axis('off')

    # ===============================================
    # PASSO 6: SALVATAGGIO AUTOMATICO
    # ===============================================
    cartella_destinazione = os.path.dirname(percorso_output)
    if not os.path.exists(cartella_destinazione):
        os.makedirs(cartella_destinazione)
        print(f"📁 Creata cartella: {cartella_destinazione}")

    print(f"💾 Salvataggio in corso...")
    plt.savefig(percorso_output, dpi=300, bbox_inches='tight')
    
    plt.close() 

    fine = time.time()
    print(f"✅ Grafo salvato con successo in: {percorso_output}")
    print(f"⏱️ Tempo impiegato: {fine - inizio:.2f} secondi!\n")


FILE_INPUT_S2 = '../DataSet_CasoStudio2/Rete_100/dataset_scenario2_cs2.csv'
FILE_OUTPUT_S2 = '../Grafi_CasoStudio2/Grafo_Scenario2_cs2.png'

genera_grafo_scenario2_cs2(FILE_INPUT_S2, FILE_OUTPUT_S2)
```

```python
import os
import time
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

def genera_grafo_scenario3_cs2(percorso_input, percorso_output):
    print(f"⚙️ Inizio generazione grafo: Scenario 3 (Ecosistema Organico E-commerce)")
    inizio = time.time()
    
    # ===============================================
    # PASSO 1: CARICAMENTO DATASET E CREAZIONE GRAFO
    # ===============================================
    print(f"📥 Caricamento del dataset da: {percorso_input}...")
    df = pd.read_csv(percorso_input)
    
    # Crea il grafo orientato
    G = nx.from_pandas_edgelist(df, 'Source', 'Target', create_using=nx.DiGraph())

    # ===============================================
    # PASSO 2: DEFINIZIONE DEI GRUPPI STRATEGICI
    # ===============================================
    print("🧠 Classificazione dei nodi secondo le logiche di tesi...")
    best_sellers = [0, 1, 2]
    
    # Micro-Hub aggiornati secondo il dataset Caso Studio 2
    micro_hubs = [10, 35, 60, 85]
    
    nodi_speciali = best_sellers + micro_hubs
    
    # Tutti gli altri nodi sono prodotti correlati raggruppati in cluster merceologici
    prodotti_catalogo = [n for n in G.nodes() if n not in nodi_speciali]

    # ===============================================
    # PASSO 3: CALCOLO DEL LAYOUT SPAZIALE
    # ===============================================
    print("🌐 Calcolo della disposizione spaziale (Layout espanso)...")
    # MODIFICA: Tela più grande
    plt.figure(figsize=(20, 15)) 
    
    pos = nx.spring_layout(G, k=0.6, iterations=100, seed=42) 

    # ===============================================
    # PASSO 4: DISEGNO DEI NODI E DEGLI ARCHI 
    # ===============================================
    print("🎨 Disegno della rete (Colori scuri e Dimensioni ingrandite)...")
    
    # MODIFICA: Nodi raddoppiati e colori proiettore
    nx.draw_networkx_nodes(G, pos, nodelist=best_sellers, node_color='#CC0000', node_size=1800, label='Best-Seller (0, 1, 2)')
    nx.draw_networkx_nodes(G, pos, nodelist=micro_hubs, node_color='#B8860B', node_size=1200, label='Micro-Hub (10, 35, 60, 85)')
    nx.draw_networkx_nodes(G, pos, nodelist=prodotti_catalogo, node_color='#005580', node_size=150, alpha=0.85, label='Prodotti Correlati (Cluster Merceologici)')

    # MODIFICA: Archi più spessi e frecce più grandi
    nx.draw_networkx_edges(G, pos, alpha=0.4, edge_color='#666666', width=1.5, arrows=True, arrowsize=20)

    # ===============================================
    # PASSO 5: ETICHETTE E DETTAGLI ESTETICI
    # ===============================================
    # MODIFICA: Font grande per i numeri
    labels = {n: str(n) for n in nodi_speciali}
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=16, font_weight='bold', font_color='white')

    # MODIFICA: Titolo più grande
    plt.title("Caso di Studio 2: Ecosistema Organico e Categorie Merceologiche (Scenario 3)", fontsize=22, fontweight='bold', pad=20)
    
    # --- MODIFICA CHIAVE PER NON SOVRAPPORRE LA LEGENDA ---
    # MODIFICA: Legenda spinta FUORI dal grafo a destra (1.02)
    # E leggermente rimpicciolita (fontsize=18, markerscale=0.8) per non dominare l'immagine.
    plt.legend(
        scatterpoints=1, 
        bbox_to_anchor=(1.02, 0.0), # <-- Valore X portato a 1.02 per metterla FUORI a destra
        loc='lower left', 
        fontsize=18, 
        markerscale=0.8, 
        framealpha=1,
        borderpad=1.0,
        labelspacing=1.0,
        handletextpad=1.0
    )
    # --------------------------------------------------------
    
    plt.margins(0)
    plt.axis('off')

    # ===============================================
    # PASSO 6: SALVATAGGIO AUTOMATICO
    # ===============================================
    cartella_destinazione = os.path.dirname(percorso_output)
    if not os.path.exists(cartella_destinazione):
        os.makedirs(cartella_destinazione)
        print(f"📁 Creata cartella: {cartella_destinazione}")

    print(f"💾 Salvataggio in corso...")
    plt.savefig(percorso_output, dpi=300, bbox_inches='tight')
    
    plt.close() 

    fine = time.time()
    print(f"✅ Grafo salvato con successo in: {percorso_output}")
    print(f"⏱️ Tempo impiegato: {fine - inizio:.2f} secondi!\n")


FILE_INPUT_S3 = '../DataSet_CasoStudio2/Rete_100/dataset_scenario3_cs2.csv'
FILE_OUTPUT_S3 = '../Grafi_CasoStudio2/Grafo_Scenario3_cs2_Proiettore.png'

genera_grafo_scenario3_cs2(FILE_INPUT_S3, FILE_OUTPUT_S3)
```
