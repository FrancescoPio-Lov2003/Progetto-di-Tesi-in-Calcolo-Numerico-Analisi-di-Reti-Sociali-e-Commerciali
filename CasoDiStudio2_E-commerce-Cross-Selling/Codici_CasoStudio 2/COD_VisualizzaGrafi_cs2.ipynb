```python
import os
import time
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# =====================================================================
# GENERAZIONE GRAFO E-COMMERCE: SCENARIO 1 (ACQUISTI DIRETTI)
# =====================================================================

def genera_grafo_scenario1_cs2(percorso_input, percorso_output):
    """
    Carica il dataset della rete e-commerce, calcola il layout spaziale tramite algoritmo 
    Force-Directed (Spring) e renderizza il grafo evidenziando i nodi 'Best-Seller' 
    rispetto ai 'Prodotti Isolati'. Ottimizzato per presentazioni e analisi visiva.
    """
    print("Inizio generazione grafo: Scenario 1 (E-commerce - Acquisti Diretti)")
    inizio = time.time()
    
    # 1. Caricamento della rete dal dataset e inizializzazione del grafo diretto
    print(f"Caricamento del dataset da: {percorso_input}...")
    df = pd.read_csv(percorso_input)
    
    G = nx.from_pandas_edgelist(df, 'Source', 'Target', create_using=nx.DiGraph())
    G.remove_edges_from(list(nx.selfloop_edges(G)))

    # 2. Classificazione topologica dei nodi
    print("Classificazione dei nodi...")
    best_sellers = [0, 1, 2]
    nodi_speciali = best_sellers
    prodotti_isolati = [n for n in G.nodes() if n not in nodi_speciali]

    # 3. Calcolo della disposizione spaziale (Spring Layout)
    print("Calcolo della disposizione spaziale...")
    plt.figure(figsize=(20, 15)) 
    pos = nx.spring_layout(G, k=0.9, iterations=100, seed=42) 

    # 4. Rendering grafico della rete (Nodi, Archi e Etichette)
    print("Disegno della rete...")
    nx.draw_networkx_nodes(G, pos, nodelist=best_sellers, node_color='#CC0000', node_size=1800, label='Best-Seller (0, 1, 2)')
    nx.draw_networkx_nodes(G, pos, nodelist=prodotti_isolati, node_color='#003366', node_size=150, alpha=0.85, label='Prodotti Isolati (Acquisti Diretti)')

    nx.draw_networkx_edges(G, pos, alpha=0.4, edge_color='#666666', width=1.5, arrows=True, arrowsize=20)

    labels = {n: str(n) for n in nodi_speciali}
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=16, font_weight='bold', font_color='white')

    # 5. Formattazione del layout globale e della legenda
    plt.title("Caso di Studio 2: Acquisti Diretti e Prodotti Isolati (Scenario 1)", fontsize=22, fontweight='bold', pad=20)
    
    plt.legend(
        scatterpoints=1, 
        bbox_to_anchor=(0.75, 0.05), 
        loc='lower left', 
        fontsize=20, 
        markerscale=0.9, 
        framealpha=1,
        borderpad=1.0,       
        labelspacing=1.0,    
        handletextpad=1.0    
    )
    
    plt.margins(0)
    plt.axis('off')

    # 6. Esportazione strutturata del file grafico
    cartella_destinazione = os.path.dirname(percorso_output)
    if not os.path.exists(cartella_destinazione):
        os.makedirs(cartella_destinazione)

    print("Salvataggio in corso...")
    plt.savefig(percorso_output, dpi=300, bbox_inches='tight')
    plt.close() 

    fine = time.time()
    print(f"Grafo salvato con successo in: {percorso_output}")
    print(f"Tempo impiegato: {fine - inizio:.2f} secondi!\n")


# =====================================================================
# ESECUZIONE SCRIPT
# =====================================================================
if __name__ == "__main__":
    FILE_INPUT_S1 = '../DataSet_CasoStudio2/Rete_100/dataset_scenario1_cs2.csv'
    FILE_OUTPUT_S1 = '../Grafi_CasoStudio2/Grafo_Scenario1_cs2.png'

    genera_grafo_scenario1_cs2(FILE_INPUT_S1, FILE_OUTPUT_S1)
```

```python
import os
import time
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# =====================================================================
# GENERAZIONE GRAFO E-COMMERCE: SCENARIO 2 (RETE MISTA E PRODOTTI HUB)
# =====================================================================

def genera_grafo_scenario2_cs2(percorso_input, percorso_output):
    """
    Carica il dataset della rete e-commerce, calcola il layout spaziale tramite algoritmo 
    Force-Directed (Spring) e renderizza il grafo evidenziando 'Best-Seller', 
    'Prodotti Hub', 'Prodotti Correlati' e 'Prodotti Isolati'. 
    Ottimizzato per presentazioni e analisi visiva.
    """
    print("Inizio generazione grafo: Scenario 2 (Rete Mista e Prodotti Hub)")
    inizio = time.time()
    
    # 1. Caricamento della rete dal dataset e inizializzazione del grafo diretto
    print(f"Caricamento del dataset da: {percorso_input}...")
    df = pd.read_csv(percorso_input)
    
    G = nx.from_pandas_edgelist(df, 'Source', 'Target', create_using=nx.DiGraph())
    G.remove_edges_from(list(nx.selfloop_edges(G)))

    # 2. Classificazione topologica dei nodi
    print("Classificazione dei nodi...")
    best_sellers = [0, 1, 2]
    hubs = [60, 70, 80, 90]
    
    prodotti_isolati = [n for n in G.nodes() if 3 <= n <= 50]
    prodotti_correlati = [n for n in G.nodes() if n >= 51 and n not in hubs]
    
    nodi_speciali = best_sellers + hubs

    # 3. Calcolo della disposizione spaziale (Spring Layout)
    print("Calcolo della disposizione spaziale...")
    plt.figure(figsize=(20, 15)) 
    pos = nx.spring_layout(G, k=0.85, iterations=120, seed=42) 

    # 4. Rendering grafico della rete (Nodi, Archi e Etichette)
    print("Disegno della rete...")
    nx.draw_networkx_nodes(G, pos, nodelist=best_sellers, node_color='#CC0000', node_size=1800, label='Best-Seller (0, 1, 2)')
    nx.draw_networkx_nodes(G, pos, nodelist=hubs, node_color='#CC7000', node_size=1400, label='Prodotti Hub / Civetta (60, 70, 80, 90)')
    nx.draw_networkx_nodes(G, pos, nodelist=prodotti_correlati, node_color='#006600', node_size=150, alpha=0.85, label='Prodotti Correlati (51-99)')
    nx.draw_networkx_nodes(G, pos, nodelist=prodotti_isolati, node_color='#003366', node_size=150, alpha=0.85, label='Prodotti Isolati (3-50)')

    nx.draw_networkx_edges(G, pos, alpha=0.4, edge_color='#666666', width=1.5, arrows=True, arrowsize=20)

    labels = {n: str(n) for n in nodi_speciali}
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=16, font_weight='bold', font_color='white')

    # 5. Formattazione del layout globale e della legenda
    plt.title("Caso di Studio 2: Rete Mista e Prodotti Hub (Scenario 2)", fontsize=22, fontweight='bold', pad=20)
    
    plt.legend(
        scatterpoints=1, 
        bbox_to_anchor=(1.02, 0.0),  
        loc='lower left',            
        fontsize=18,                 
        markerscale=0.8,             
        framealpha=1,
        borderpad=1.0,
        labelspacing=1.0,
        handletextpad=1.0
    )
    
    plt.margins(0)
    plt.axis('off')

    # 6. Esportazione strutturata del file grafico
    cartella_destinazione = os.path.dirname(percorso_output)
    if not os.path.exists(cartella_destinazione):
        os.makedirs(cartella_destinazione)

    print("Salvataggio in corso...")
    plt.savefig(percorso_output, dpi=300, bbox_inches='tight')
    plt.close() 

    fine = time.time()
    print(f"Grafo salvato con successo in: {percorso_output}")
    print(f"Tempo impiegato: {fine - inizio:.2f} secondi!\n")


# =====================================================================
# ESECUZIONE SCRIPT
# =====================================================================
if __name__ == "__main__":
    FILE_INPUT_S2 = '../DataSet_CasoStudio2/Rete_100/dataset_scenario2_cs2.csv'
    FILE_OUTPUT_S2 = '../Grafi_CasoStudio2/Grafo_Scenario2_cs2.png'

    genera_grafo_scenario2_cs2(FILE_INPUT_S2, FILE_OUTPUT_S2)
```
