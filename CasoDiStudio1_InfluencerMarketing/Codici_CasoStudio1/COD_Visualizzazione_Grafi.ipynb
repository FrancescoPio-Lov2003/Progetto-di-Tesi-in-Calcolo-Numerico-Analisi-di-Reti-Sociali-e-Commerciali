```python
import os
import time
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# =====================================================================
# GENERAZIONE GRAFO: SCENARIO 1 (INFLUENCER MARKETING)
# =====================================================================

def genera_grafo_scenario1(percorso_input, percorso_output):
    """
    Carica il dataset della rete, calcola il layout spaziale tramite algoritmo 
    Force-Directed (Spring) e renderizza il grafo evidenziando i nodi 
    'Influencer' rispetto ai 'Bot/Follower'. Ottimizzato per presentazioni.
    """
    print("Inizio generazione grafo: Scenario 1 (Influencer Marketing)")
    inizio = time.time()
    
    # 1. Caricamento della rete dal dataset e inizializzazione del grafo diretto
    print(f"Caricamento del dataset da: {percorso_input}...")
    df = pd.read_csv(percorso_input)
    
    G = nx.from_pandas_edgelist(df, 'Source', 'Target', create_using=nx.DiGraph())
    G.remove_edges_from(list(nx.selfloop_edges(G)))

    # 2. Classificazione topologica dei nodi
    print("Classificazione dei nodi...")
    influencers = [0, 1, 2]
    nodi_normali = [n for n in G.nodes() if n not in influencers]

    # 3. Calcolo della disposizione spaziale (Spring Layout)
    print("Calcolo della disposizione spaziale...")
    plt.figure(figsize=(20, 15)) 
    pos = nx.spring_layout(G, k=0.9, iterations=100, seed=42) 

    # 4. Rendering grafico della rete (Nodi, Archi e Etichette)
    print("Disegno della rete...")
    nx.draw_networkx_nodes(G, pos, nodelist=influencers, node_color='#CC0000', node_size=1800, label='Influencer (0, 1, 2)')
    nx.draw_networkx_nodes(G, pos, nodelist=nodi_normali, node_color='#003366', node_size=150, alpha=0.85, label='Bot / Follower Inattivi')

    nx.draw_networkx_edges(G, pos, alpha=0.4, edge_color='#666666', width=1.5, arrows=True, arrowsize=20)

    labels = {n: str(n) for n in influencers}
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=16, font_weight='bold', font_color='white')

    # 5. Formattazione del layout globale e della legenda
    plt.title("Caso di Studio 1: Influencer Marketing (Scenario 1)", fontsize=22, fontweight='bold', pad=20)
    
    plt.legend(
        scatterpoints=1, 
        bbox_to_anchor=(0.75, 0.05), 
        loc='lower left', 
        fontsize=20, 
        markerscale=0.9, 
        framealpha=1,
        borderpad=1.0,       
        labelspacing=1.0,    
        handletextpad=1.0    
    )
    
    plt.margins(0)
    plt.axis('off')

    # 6. Esportazione strutturata del file grafico
    cartella_destinazione = os.path.dirname(percorso_output)
    if not os.path.exists(cartella_destinazione):
        os.makedirs(cartella_destinazione)

    print("Salvataggio in corso...")
    plt.savefig(percorso_output, dpi=300, bbox_inches='tight')
    plt.close() 

    fine = time.time()
    print(f"Grafo salvato con successo in: {percorso_output}")
    print(f"Tempo impiegato: {fine - inizio:.2f} secondi!\n")


# =====================================================================
# ESECUZIONE SCRIPT
# =====================================================================
if __name__ == "__main__":
    FILE_INPUT = '../DataSet_CasoStudio1/Rete_100/dataset_scenario1.csv'
    FILE_OUTPUT = '../Grafi_CasoStudio1/Grafo_Scenario1.png'

    genera_grafo_scenario1(FILE_INPUT, FILE_OUTPUT)
```

```python
import os
import time
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# =====================================================================
# GENERAZIONE GRAFO: SCENARIO 2 (RETE MISTA E HUB STRATEGICI)
# =====================================================================

def genera_grafo_scenario2(percorso_input, percorso_output):
    """
    Carica il dataset della rete, calcola il layout spaziale tramite algoritmo 
    Force-Directed (Spring) e renderizza il grafo evidenziando i nodi 
    'Influencer', 'Hub Strategici', 'Utenti Attivi' e 'Bot'.
    Ottimizzato per presentazioni e analisi visiva.
    """
    print("Inizio generazione grafo: Scenario 2 (Rete Mista e Hub)")
    inizio = time.time()
    
    # 1. Caricamento della rete dal dataset e inizializzazione del grafo diretto
    print(f"Caricamento del dataset da: {percorso_input}...")
    df = pd.read_csv(percorso_input)
    
    G = nx.from_pandas_edgelist(df, 'Source', 'Target', create_using=nx.DiGraph())
    G.remove_edges_from(list(nx.selfloop_edges(G)))

    # 2. Classificazione topologica dei nodi
    print("Classificazione dei nodi...")
    influencers = [0, 1, 2]
    hubs = [60, 70, 80, 90]
    bots = [n for n in G.nodes() if 3 <= n <= 50]
    utenti_attivi = [n for n in G.nodes() if n >= 51 and n not in hubs]
    nodi_speciali = influencers + hubs

    # 3. Calcolo della disposizione spaziale (Spring Layout)
    print("Calcolo della disposizione spaziale...")
    plt.figure(figsize=(20, 15)) 
    pos = nx.spring_layout(G, k=0.6, iterations=100, seed=42) 

    # 4. Rendering grafico della rete (Nodi, Archi e Etichette)
    print("Disegno della rete...")
    nx.draw_networkx_nodes(G, pos, nodelist=influencers, node_color='#CC0000', node_size=1800, label='Influencer (0, 1, 2)')
    nx.draw_networkx_nodes(G, pos, nodelist=hubs, node_color='#CC7000', node_size=1400, label='Hub Strategici (60, 70, 80, 90)')
    nx.draw_networkx_nodes(G, pos, nodelist=utenti_attivi, node_color='#006600', node_size=150, alpha=0.85, label='Utenti Attivi (51-99)')
    nx.draw_networkx_nodes(G, pos, nodelist=bots, node_color='#003366', node_size=150, alpha=0.85, label='Bot Inattivi (3-50)')

    nx.draw_networkx_edges(G, pos, alpha=0.4, edge_color='#666666', width=1.5, arrows=True, arrowsize=20)

    labels = {n: str(n) for n in nodi_speciali}
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=16, font_weight='bold', font_color='white')

    # 5. Formattazione del layout globale e della legenda
    plt.title("Caso di Studio 1: Rete Mista e Hub Strategici (Scenario 2)", fontsize=22, fontweight='bold', pad=20)
    
    plt.legend(
        scatterpoints=1, 
        bbox_to_anchor=(1.02, 0.0), 
        loc='lower left',            
        fontsize=18,                 
        markerscale=0.8,             
        framealpha=1,
        borderpad=1.0,
        labelspacing=1.0,
        handletextpad=1.0
    )
    
    plt.margins(0)
    plt.axis('off')

    # 6. Esportazione strutturata del file grafico
    cartella_destinazione = os.path.dirname(percorso_output)
    if not os.path.exists(cartella_destinazione):
        os.makedirs(cartella_destinazione)

    print("Salvataggio in corso...")
    plt.savefig(percorso_output, dpi=300, bbox_inches='tight')
    plt.close() 

    fine = time.time()
    print(f"Grafo salvato con successo in: {percorso_output}")
    print(f"Tempo impiegato: {fine - inizio:.2f} secondi!\n")


# =====================================================================
# ESECUZIONE SCRIPT
# =====================================================================
if __name__ == "__main__":
    FILE_INPUT = '../DataSet_CasoStudio1/Rete_100/dataset_scenario2.csv'
    FILE_OUTPUT = '../Grafi_CasoStudio1/Grafo_Scenario2.png'

    genera_grafo_scenario2(FILE_INPUT, FILE_OUTPUT)
```

```python
import os
import time
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

# =====================================================================
# GENERAZIONE GRAFO: SCENARIO 3 (TOPOLOGIA SMALL-WORLD)
# =====================================================================

def genera_grafo_scenario_3(percorso_input, percorso_output):
    """
    Carica il dataset della rete, calcola il layout spaziale tramite algoritmo 
    Force-Directed (Spring) e renderizza il grafo evidenziando i nodi 
    'Influencer', 'Micro-Hub' e 'Follower Organici' in un cluster Small-World.
    Ottimizzato per presentazioni e analisi visiva.
    """
    print("Inizio generazione grafo: Scenario 3 (Topologia Small-World)")
    inizio = time.time()
    
    # 1. Caricamento della rete dal dataset e inizializzazione del grafo diretto
    print(f"Caricamento del dataset da: {percorso_input}...")
    df = pd.read_csv(percorso_input)
    
    G = nx.from_pandas_edgelist(df, 'Source', 'Target', create_using=nx.DiGraph())

    # 2. Classificazione topologica dei nodi
    print("Classificazione dei nodi...")
    influencers = [0, 1, 2]
    micro_hubs = [10, 15, 25, 45, 55, 80]
    nodi_speciali = influencers + micro_hubs
    follower_organici = [n for n in G.nodes() if n not in nodi_speciali]

    # 3. Calcolo della disposizione spaziale (Spring Layout)
    print("Calcolo della disposizione spaziale...")
    plt.figure(figsize=(20, 15)) 
    pos = nx.spring_layout(G, k=0.6, iterations=100, seed=42) 

    # 4. Rendering grafico della rete (Nodi, Archi e Etichette)
    print("Disegno della rete...")
    nx.draw_networkx_nodes(G, pos, nodelist=influencers, node_color='#CC0000', node_size=1800, label='Influencer (0, 1, 2)')
    nx.draw_networkx_nodes(G, pos, nodelist=micro_hubs, node_color='#B8860B', node_size=1200, label='Micro-Hub (10, 15, 25, 45, 55, 80)')
    nx.draw_networkx_nodes(G, pos, nodelist=follower_organici, node_color='#005580', node_size=150, alpha=0.85, label='Follower Organici (Cluster)')

    nx.draw_networkx_edges(G, pos, alpha=0.4, edge_color='#666666', width=1.5, arrows=True, arrowsize=20)

    labels = {n: str(n) for n in nodi_speciali}
    nx.draw_networkx_labels(G, pos, labels=labels, font_size=16, font_weight='bold', font_color='white')

    # 5. Formattazione del layout globale e della legenda
    plt.title("Caso di Studio 1: Community Organica 100% Attiva (Scenario 3)", fontsize=22, fontweight='bold', pad=20)
    
    plt.legend(
        scatterpoints=1, 
        bbox_to_anchor=(0.75, 0.05), 
        loc='lower left', 
        fontsize=20, 
        markerscale=0.9, 
        framealpha=1,
        borderpad=1.0,
        labelspacing=1.0,
        handletextpad=1.0
    )
    
    plt.margins(0)
    plt.axis('off')

    # 6. Esportazione strutturata del file grafico
    cartella_destinazione = os.path.dirname(percorso_output)
    if not os.path.exists(cartella_destinazione):
        os.makedirs(cartella_destinazione)

    print("Salvataggio in corso...")
    plt.savefig(percorso_output, dpi=300, bbox_inches='tight')
    plt.close() 

    fine = time.time()
    print(f"Grafo salvato con successo in: {percorso_output}")
    print(f"Tempo impiegato: {fine - inizio:.2f} secondi!\n")


# =====================================================================
# ESECUZIONE SCRIPT
# =====================================================================
if __name__ == "__main__":
    INPUT_CSV = '../DataSet_CasoStudio1/Rete_100/dataset_scenario3.csv'
    OUTPUT_PNG = '../Grafi_CasoStudio1/Grafo_Scenario3.png'

    genera_grafo_scenario_3(INPUT_CSV, OUTPUT_PNG)
```
